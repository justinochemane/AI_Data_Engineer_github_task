```python
# [Problem 1] Data acquisition
```

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load the iris dataset
iris = load_iris()

# Creating the dataFrames
X = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print("X DataFrame:")
print(X.head())
```

```python
#. Create the y DataFrame (Objective Variable)
```

```python
y = pd.DataFrame(data=iris.target, columns=['Species'])
print("\ny DataFrame:")
print(y.head())
```

```python
# [Problem 2] Combining data
```

```python
# Combine X and y into a single DataFrame df
df = pd.concat([X, y], axis=1)

# Display the first few rows of the combined DataFrame
print(df.head())
```

```python
# [Problem 6] Creating a diagram
```

```python
# Set up matplotlib and seaborn styles for better visuals
plt.style.use('ggplot')
sns.set_style("whitegrid")
```

```python
# 1. Pie chart of the number of samples per label
```

```python
species_counts = df['Species'].value_counts()
species_names = iris.target_names

plt.figure(figsize=(8, 8))
plt.pie(species_counts, labels=species_names, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Proportion of Samples per Iris Species')
plt.show()
```

```python
# 2. Box plots for each feature
```

```python
features = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']
for feature in features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Species', y=feature, data=df)
    plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
    plt.title(f'Box Plot of {feature} by Species')
    plt.show()
```

```python
# 3. Violin plots for each feature
```

```python
# [Problem 7] Confirming the relationship between features
```

```python
## Scatter plots for all combinations of two features
```

```python
from itertools import combinations
features_to_plot = df.columns[:-1]
feature_combinations = list(combinations(features_to_plot, 2))
for feature1, feature2 in feature_combinations:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(data=df, x=feature1, y=feature2, hue='Species', palette='viridis')
    plt.title(f'Scatter Plot of {feature1} vs {feature2}')
    plt.show()
```

```python
## Scatterplot matrix (pair plot)
```

```python
sns_plot = sns.pairplot(df, hue='Species', palette='viridis')
sns_plot.fig.suptitle('Scatter Plot Matrix of All Features', y=1.02)
plt.show()
```

```python
# Correlation coefficient matrix
```

```python
correlation_matrix = df[features_to_plot].corr()
print("\nCorrelation Coefficient Matrix:")
print(correlation_matrix)
```

```python
# Heatmap of the correlation coefficient matrix
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap of Feature Correlation Matrix')
plt.show()
```

```python
# [Problem 8] Explaining the results of visualized graphs and tables
```

From the visualizations and tables, we can draw several conclusions about the Iris dataset and the relationships between its features.

### Feature Distributions and Separability
* The **pie chart** confirms the dataset is perfectly **balanced**, with 50 samples for each of the three species: setosa, versicolor, and virginica. This is important as it prevents a machine learning model from being biased towards one class.
* The **box plots and violin plots** reveal significant differences in feature distributions across the species. **Iris setosa** is clearly separated from the other two species, particularly by its **petal length** and **petal width**, which are much smaller. The other two species, **versicolor** and **virginica**, have more overlap, especially in sepal measurements, but can still be distinguished by their petal measurements, with virginica generally having larger petals.

### Feature Relationships
* The **scatter plots and pair plot** visually demonstrate the relationships between all pairs of features.
    * **Iris setosa** (purple points) forms a distinct, tight cluster in almost all plots, indicating it's easily separable from the other species.
    * **Iris versicolor** and **Iris virginica** show some overlap, but as `petal length` and `petal width` increase, the two species tend to diverge. This is most evident in the plot of `petal length` vs. `petal width`.
* The **correlation matrix and heatmap** quantify these relationships.
    * There is a very strong positive correlation (**0.96**) between `petal length` and `petal width`, indicating these features are highly redundant.
    * `sepal length` also has a strong positive correlation with both `petal length` (**0.87**) and `petal width` (**0.82**).
    * `sepal width` has a weak correlation with the other features, and a slight negative correlation with `sepal length` (**-0.12**).

### Summary
In essence, the data reveals that `petal length` and `petal width` are the most **discriminative** features for separating the iris species, particularly for distinguishing setosa from the rest. While `sepal length` and `sepal width` provide some information, they are less effective for classification. This suggests that a machine learning model could achieve high accuracy by primarily using the petal measurements.
