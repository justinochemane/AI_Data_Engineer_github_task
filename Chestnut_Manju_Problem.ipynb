```python
def simulate_growth(
    initial_volume: float, 
    target_volume: float, 
    doubling_time: float = 5, 
    num_points: int = 1000
) -> tuple[np.ndarray, np.ndarray, float]:
    """Simulate volume growth over time and return time/volume arrays and total time."""
    total_time = calculate_growth_time(initial_volume, target_volume, doubling_time)
    time_points = np.linspace(0, total_time, num_points)
    volumes = initial_volume * (2 ** (time_points / doubling_time))
    return time_points, volumes, total_time
```

```python
def plot_growth(
    time_points: np.ndarray, 
    volumes: np.ndarray, 
    target_volume: float, 
    initial_volume: float, 
    total_time: float
) -> None:
    """Plot volume growth over time with logarithmic scale and annotations."""
    plt.figure(figsize=(10, 6))
    plt.plot(time_points, volumes, label="Volume Growth", color="blue")
    
    # Highlight target and initial volumes
    plt.axhline(target_volume, color="red", linestyle="--", label="Target Volume")
    plt.axhline(initial_volume, color="green", linestyle="--", label="Initial Volume")
    
    plt.yscale("log")
    plt.xlabel("Time (minutes)")
    plt.ylabel("Volume (log scale)")
    plt.title(f"Exponential Growth: Fills Target in {total_time:.2f} minutes")
    plt.legend()
    plt.grid(True, which="both", linestyle="--")
    plt.show()
```

```python
# Example 1: Chestnut Bun vs. Solar System
if __name__ == "__main__":
    # Define volumes (ensure consistent units, e.g., cubic meters)
    chestnut_bun_volume = 1178.5e-6  # Convert cm³ to m³ (0.0011785 m³)
    solar_system_volume = 3.82e29 * 1e9  # Convert km³ to m³ (3.82e38 m³)
    
    # Calculate time to fill Solar System
    time_solar = calculate_growth_time(chestnut_bun_volume, solar_system_volume)
    print(
        f"Time to fill Solar System: {time_solar:.2f} minutes "
        f"(~{time_solar / 60:.1f} hours)"
    )
    
    # Simulate and plot growth
    t, v, t_total = simulate_growth(chestnut_bun_volume, solar_system_volume)
    plot_growth(t, v, solar_system_volume, chestnut_bun_volume, t_total)
```

```python
# Example 2: Ball vs. Tokyo Dome
ball_volume = 1.0  # m³ (e.g., a large ball)
tokyo_dome_volume = 1.24e6  # m³ (official volume: ~1.24 million m³)
    
# Calculate time to fill Tokyo Dome
time_dome = calculate_growth_time(ball_volume, tokyo_dome_volume)
print(
        f"\nTime to fill Tokyo Dome: {time_dome:.2f} minutes "
        f"(~{time_dome / 60:.2f} hours)"
    )
    
# Simulate and plot
t_dome, v_dome, t_total_dome = simulate_growth(ball_volume, tokyo_dome_volume)
plot_growth(t_dome, v_dome, tokyo_dome_volume, ball_volume, t_total_dome)
```
