```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
import numpy as np

# --- 1. Data Loading and Preprocessing ---

# Load the Iris Dataset
iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names

# Filter for Binary Classification (Versicolor: index 1, Virginica: index 2)
# The features (X_binary) for these two classes
binary_indices = np.where((y == 1) | (y == 2))
X_binary = X[binary_indices]
y_binary = y[binary_indices]

# Encode Labels: Map the two classes (1 and 2) to (0 and 1)
# This is required for 'binary_crossentropy' loss
y_encoded = y_binary - 1

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(
    X_binary, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print(f"Dataset size: {len(X_binary)} samples (Versicolor & Virginica)")
print(f"Training set shape: {X_train.shape}")
print("-" * 40)

# --- 2. Keras Model Definition (Sequential API) ---

# Define the model architecture using the Sequential API
model_iris_binary = tf.keras.Sequential([
    # Input and Hidden Layer: 4 input features, 10 neurons, ReLU activation
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    
    # Output Layer: 1 neuron, Sigmoid activation for binary classification
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Display the model structure
model_iris_binary.summary()
print("-" * 40)

# --- 3. Compile the Model ---

# Configure the model's learning process
model_iris_binary.compile(
    optimizer='adam',
    # Use binary_crossentropy for binary classification
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# --- 4. Train (Fit) the Model ---

print("Starting model training (50 epochs)...")
history = model_iris_binary.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=4,
    verbose=0 # Set to 1 for progress bar
)
print("Training finished.")

# --- 5. Evaluate the Model ---

loss, accuracy = model_iris_binary.evaluate(X_test, y_test, verbose=0)
print(f"\nModel Evaluation on Test Data:")
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
# [Problem 4] Learn Iris (multi-level classification) with Keras
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from tensorflow.keras.utils import to_categorical

# --- 1. Data Loading and Preprocessing ---

# Load the Iris Dataset (all three classes: 0, 1, 2)
iris = load_iris()
X = iris.data
y = iris.target # y now contains 0, 1, and 2

# One-Hot Encode Labels
# Convert integer labels (0, 1, 2) into vectors (e.g., [1, 0, 0], [0, 1, 0], [0, 0, 1])
y_one_hot = to_categorical(y, num_classes=3)

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y_one_hot, test_size=0.2, random_state=42, stratify=y
)

print(f"Dataset size: {len(X)} samples (Setosa, Versicolor, & Virginica)")
print(f"Training set shape (Features): {X_train.shape}")
print(f"Training label shape (One-Hot Encoded): {y_train.shape}")
print("-" * 55)

# --- 2. Keras Model Definition (Functional API) ---

# Define the input layer with 4 features
input_layer = tf.keras.Input(shape=(4,))

# Hidden Layer: 10 neurons with ReLU activation
hidden_layer = tf.keras.layers.Dense(10, activation='relu')(input_layer)

# Output Layer: 3 neurons (for the 3 classes) with Softmax activation
# Softmax ensures the output probabilities sum to 1.
output_layer = tf.keras.layers.Dense(3, activation='softmax')(hidden_layer)

# Create the final model object
model_iris_multi = tf.keras.Model(inputs=input_layer, outputs=output_layer)

# Display the model structure
model_iris_multi.summary()
print("-" * 55)

# --- 3. Compile the Model ---

# Configure the model's learning process
model_iris_multi.compile(
    optimizer='adam',
    # Use categorical_crossentropy for multi-class classification with one-hot labels
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# --- 4. Train (Fit) the Model ---

print("Starting model training (50 epochs)...")
history = model_iris_multi.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=4,
    verbose=0
)
print("Training finished.")

# --- 5. Evaluate the Model ---

loss, accuracy = model_iris_multi.evaluate(X_test, y_test, verbose=0)
print(f"\nModel Evaluation on Test Data:")
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import fetch_california_housing
import numpy as np

# --- 1. Data Loading and Preprocessing ---

# Load the California Housing dataset (a standard, real-world regression dataset)
housing = fetch_california_housing()
X = housing.data
y = housing.target

# Standardize the features (Good practice for most deep learning models)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# X_train.shape[1] is the number of features in the dataset (8 for California Housing)
input_features = X_train.shape[1]

print(f"Dataset features: {input_features}")
print(f"Training set shape: {X_train.shape}")
print("-" * 40)

# --- 2. Keras Model Definition (Sequential API) ---

# Define the model architecture
model_house_prices = tf.keras.Sequential([
    # Input and Hidden Layer 1
    # input_shape must match the number of features
    tf.keras.layers.Dense(64, activation='relu', input_shape=(input_features,)),

    # Hidden Layer 2
    tf.keras.layers.Dense(64, activation='relu'),

    # Output Layer: 1 neuron with no activation (linear output)
    # This layer predicts the continuous house price value.
    tf.keras.layers.Dense(1)
])

# Display the model structure
model_house_prices.summary()
print("-" * 40)

# --- 3. Compile the Model ---

# Configure the model's learning process
model_house_prices.compile(
    optimizer='adam',
    # Use Mean Squared Error (MSE) for regression tasks
    loss='mse',
    # Use Mean Absolute Error (MAE) as a key metric for interpretability
    metrics=['mae']
)

# --- 4. Train (Fit) the Model ---

print("Starting model training (100 epochs)...")
history = model_house_prices.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    verbose=0 # Set to 1 for progress bar
)
print("Training finished.")

# --- 5. Evaluate the Model ---

loss, mae = model_house_prices.evaluate(X_test, y_test, verbose=0)
print(f"\nModel Evaluation on Test Data:")
print(f"Loss (MSE): {loss:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
```

```python
# [Problem 6] Learning MNIST with Keras
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# --- 1. Data Loading and Preprocessing ---

# Load the MNIST dataset (images and labels)
(X_train_raw, y_train_raw), (X_test_raw, y_test_raw) = mnist.load_data()

# Reshape the data for CNN: (samples, height, width, channels)
# MNIST images are 28x28 grayscale, so channels=1
X_train = X_train_raw.reshape(-1, 28, 28, 1).astype('float32')
X_test = X_test_raw.reshape(-1, 28, 28, 1).astype('float32')

# Normalize the pixel values from 0-255 to 0-1
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-Hot Encode Labels (10 classes: 0 through 9)
y_train = to_categorical(y_train_raw, num_classes=10)
y_test = to_categorical(y_test_raw, num_classes=10)

input_shape = X_train.shape[1:] # (28, 28, 1)

print(f"Input image shape: {input_shape}")
print(f"Training set size: {len(X_train)} samples")
print("-" * 55)

# --- 2. Keras Model Definition (Convolutional Neural Network) ---

model_mnist = tf.keras.Sequential([
    # 1st Convolutional Block: 32 filters, 3x3 kernel
    tf.keras.layers.Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=input_shape),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)), # Downsample by 2x2

    # 2nd Convolutional Block: 64 filters, 3x3 kernel
    tf.keras.layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),

    # Flatten the 3D output to 1D for the Dense layers
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dropout(0.5), # Dropout for regularization

    # Dense Hidden Layer
    tf.keras.layers.Dense(128, activation='relu'),

    # Output Layer: 10 neurons (for 10 digits) with Softmax activation
    tf.keras.layers.Dense(10, activation='softmax')
])

# Display the model structure
model_mnist.summary()
print("-" * 55)

# --- 3. Compile the Model ---

model_mnist.compile(
    optimizer='adam',
    # Use categorical_crossentropy for multi-class classification with one-hot labels
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# --- 4. Train (Fit) the Model ---

print("Starting model training (10 epochs)...")
history = model_mnist.fit(
    X_train,
    y_train,
    epochs=10, # Fewer epochs are often enough for MNIST with a CNN
    batch_size=128,
    validation_data=(X_test, y_test),
    verbose=0
)
print("Training finished.")

# --- 5. Evaluate the Model ---

loss, accuracy = model_mnist.evaluate(X_test, y_test, verbose=0)
print(f"\nModel Evaluation on Test Data:")
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
## 1. Iris (Binary Classification)
## Task: Classify Iris-versicolor (0) and Iris-virginica (1).

## Data Preparation
```

```python
# Load and filter data (same as Keras setup)
iris = load_iris()
X = iris.data
y = iris.target
binary_indices = np.where((y == 1) | (y == 2))
X_binary = X[binary_indices].astype(np.float32)
y_encoded = (y[binary_indices] - 1).astype(np.float32).reshape(-1, 1)

# Split and convert to Tensors
X_train, X_test, y_train, y_test = train_test_split(X_binary, y_encoded, test_size=0.2, random_state=42)
train_data = TensorDataset(torch.from_numpy(X_train), torch.from_numpy(y_train))
train_loader = DataLoader(train_data, batch_size=4, shuffle=True)
```

```python
# PyTorch Model
```

```python
class IrisBinaryNet(nn.Module):
    def __init__(self, input_size):
        super(IrisBinaryNet, self).__init__()
        # Define layers
        self.fc1 = nn.Linear(input_size, 10) # Input: 4, Hidden: 10
        self.fc2 = nn.Linear(10, 1)        # Output: 1 for binary classification

    def forward(self, x):
        # Define forward pass
        x = torch.relu(self.fc1(x))
        x = torch.sigmoid(self.fc2(x)) # Sigmoid for probability output
        return x

# Instantiate and Train
model_binary = IrisBinaryNet(input_size=4).to(device)
criterion = nn.BCELoss() # Binary Cross-Entropy Loss
optimizer = optim.Adam(model_binary.parameters(), lr=0.01)

# Manual Training Loop (50 epochs)
for epoch in range(50):
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model_binary(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
# print(f"Iris Binary Training Loss after 50 epochs: {loss.item():.4f}")
```

Here is a summary of the differences between **TensorFlow/Keras** and **PyTorch** across the three requested dimensions.

---

## 💻 Framework Comparison: TensorFlow/Keras vs. PyTorch

| Feature | TensorFlow (Keras) | PyTorch |
| :--- | :--- | :--- |
| **Execution** | **Define-and-Run (Static Graph)**, though TF2 primarily uses **Define-by-Run (Eager)**. | **Define-by-Run (Eager Execution)** is the native and primary mode. |
| **Model Definition** | High-level, abstracted, and declarative (Sequential/Functional API). | Low-level, Pythonic, explicit class structure (`nn.Module`). |

***

### 1. Calculation Speed ⚡

| Framework | Summary | Detail |
| :--- | :--- | :--- |
| **TensorFlow/Keras** | **Highly optimized** for deployment and production; excellent large-scale distribution. | Can be marginally slower in Eager Mode than PyTorch for research due to more overhead. **Static Graphs (used for deployment)** can offer significant speed advantages in production environments (via **TensorFlow Lite/Serving** and **XLA compilation**). |
| **PyTorch** | **Excellent speed for development and research**; generally fast due to minimal overhead in Eager Mode. | Relies heavily on **JIT (Just-In-Time) compilation** using **TorchScript** to achieve production-level speeds and deployment optimizations (via **TorchServe**). Often the preferred choice for cutting-edge research models that need rapid iteration. |

***

### 2. Number of Lines of Code and Readability 📖

| Framework | Number of Lines of Code | Readability/Clarity |
| :--- | :--- | :--- |
| **TensorFlow/Keras** | **Lower** ($\downarrow$ Lines of Code) | **Higher Abstraction.** Keras models are highly readable and concise (e.g., `model.fit()` vs. a manual loop). Excellent for quickly prototyping standard models. |
| **PyTorch** | **Higher** ($\uparrow$ Lines of Code) | **Higher Explicitness.** Requires writing the model definition, dataset loading, and the full training loop explicitly. This gives **total control** but results in more boilerplate code. Very intuitive for developers familiar with pure Python/NumPy. |

***

### 3. Functions Provided 🛠️

| Framework | Core Strengths and Functions | Focus |
| :--- | :--- | :--- |
| **TensorFlow/Keras** | **Production and Deployment Ecosystem.** | **Full MLOps Pipeline.** Excellent tools for deployment across platforms: **TensorFlow Serving** (servers), **TensorFlow Lite** (mobile/IoT), and **TensorFlow.js** (web browsers). It offers a unified high-level Keras API for ease of use. |
| **PyTorch** | **Research and Customization.** | **Dynamic Graph Control.** Strong support for highly customized or experimental architectures (e.g., in NLP or complex GANs) where the computation graph changes dynamically. The **Torch Ecosystem** provides domain libraries like **TorchText** and **TorchVision**, and **TorchServe** for deployment. |
