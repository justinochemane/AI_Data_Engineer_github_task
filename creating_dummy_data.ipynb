```python
## [Problem 1] Creating random numbers[Problem 1] Creating random numbers
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set the seed for reproductibility to ensure that the random numbers generated will bbe the same every time the script is run.
np.random.seed(0)

# Define the mean of the first two-dimensional normal distribution
mean = [-3, 0]

# Define the covariance matrix
covariance = [[1.0, 0.8],
              [0.8, 1.0]]

# Generate 500 random numbers from the first multivariate normal distribution 
random_numbers = np.random.multivariate_normal(mean, covariance, 500)

# Print 500 random numbers from the first multivariate normal distribution

print(f"Shape of the generated random numbers: {random_numbers.shape}")
print("\nFirst 5 generated random numbers:")
print(random_numbers[:5])


```

```python
# [Problem 2] Visualization by scatter plotz
```

```python
# Create a scatter plot of the generated data points.
plt.figure(figsize=(8, 6)) # Set the figure size for better readability
plt.scatter(random_numbers[:, 0], random_numbers[:, 1], alpha=0.6, s=20) # 'apha' for transparency, 's' for marker size

# Add labels and a title for clarity
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.title("Scatter Plot of 2D Normally Distributed Random Numbers")
plt.grid(True) # Add a grid for easier reading values

# Display the plot 
plt.show()
```

```python
# [Problem 3] Visualization by histogram
```

```python
# Create a figure with two subplot for the histogram
plt.figure(figsize=(12, 5))

# Determine the common x-axis limits for both histograms to aligh them.
x_min = np.min(random_numbers)
x_max = np.max(random_numbers)

# Histogram for Dimension 1 (first column)
plt.subplot(1, 2, 1) # row, column, first plot
plt.hist(random_numbers[:, 0], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
plt.title("Histogram of Dimension 1")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.xlim(x_min, x_max) # Set the x-axis limits ro align with the other plot
plt.grid(axis='y', alpha=0.75)

# Histogram for Dimension 2 (second column)
plt.subplot(1, 2, 2) # row, column, second plot
plt.hist(random_numbers[:, 1], bins=30, alpha=0.7, color='lightcoral', edgecolor='black')
plt.title("Histogram of Dimension 2")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.xlim(x_min, x_max) # Set the x-axis limits ro align with the other plot
plt.grid(axis='y', alpha=0.75)

plt.tight_layout() # Adjust layout
plt.show()
```

```python
# [Problem 4] Addition of data
```

```python
# Define the mean of the second two-dimensional normal distribution
mean_problem4 = [0, -3]

# Generate 500 new radom numbers from the second multivariante normal distribution
random_numbers_problem4 = np.random.multivariate_normal(mean_problem4, covariance, 500)

# Print the shape and a few example of the new data 
print (f"\nShape of random numbers (Probem 4): {random_numbers_problem4.shape}")
print("First 5 generated random numbers (Problem 4):")
print (random_numbers_problem4[:5])

```

```python
# Combined Visualization by Scatter Plot 
# Create a single figure for the combined scatter plot
plt.figure(figsize=(10, 8))

# Plot data form Probem 1 With a legend label '0'
plt.scatter(random_numbers[:, 0], random_numbers[:, 1], alpha=0.6, s=20, label='0 (Problem 1)', color='blue')

# Plot data form Probem 4 With a legend label '1'
plt.scatter(random_numbers_problem4[:, 0], random_numbers_problem4[:, 1], alpha=0.6, s=20, label='1 (Problem 4)', color='red')

# Add labels and a title for clarity
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.title("Scatter Plot of Two 2D Normally Distributed Random Number Sets")
plt.grid(True)
plt.legend()
plt.axvline (0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline (0, color='gray', linestyle='--', linewidth=0.8)

plt.show()
```

```python
# [Problem 5] Data combination
```

```python
# Combine the two ndarrays into a single ndarray of shape (1000,2) ---> 500+500
# We use np.vstack () to vertically stack the arrays.
combined_data = np.vstack ((random_numbers, random_numbers_problem4)) # random_numbers == random_numbers_problem1

# Print the shape of the combined data to verify
print (f"\nShape of the combined data: {combined_data.shape}")
print("First 5 rows of combined data:")
print(combined_data[:5])
```

```python
# [Problem 6] Labeling
```

```python
 # Create labels dor each dataset.
# 500 labels of 0 for the first dataset (Problem 1)
labels_problem1 = np.zeros((500, 1))
# 500 labels of 1 for the first dataset (Problem 4)
labels_problem4 = np.ones((500, 1))

# Combine the data arrays with their respective label arrays
labeled_data_problem1 = np.hstack ((random_numbers, labels_problem1)) # random_numbers_problem1 == random_numbers
labeled_data_problem4 = np.hstack ((random_numbers_problem4, labels_problem4))

# Vertically stack te two labeled arrays to create a single , final dataset
labeled_data = np.vstack((labeled_data_problem1, labeled_data_problem4))

# Print the shape of the final labeled data to verify it's (1000, 3)

print(f"\nShape of the final labeled data: {labeled_data.shape}")
print("First 5 rows of final labeled data (with labels):")
print(labeled_data[:5])
print("\nRows 500-504 of finals labeled data (with labels):")
print (labeled_data[500:505])
```
